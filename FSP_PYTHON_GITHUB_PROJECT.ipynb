# Q1) Write a python program to create a DataFrame containing missing values. Perform the following operations:
a) Detect missing values  b) Replace missing values with mean/median  c) Drop rows contaoning missing values 

```python
import pandas as pd
import numpy as np

data = {
    'A': [10, 20, np.nan, 40, 50],
    'B': [5, np.nan, 15, 20, 25],
    'C': [100, 200, 300, np.nan, 500]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nMissing Values:")
print(df.isnull())

df_mean = df.fillna(df.mean())
print("\nDataFrame after replacing missing values with Mean:")
print(df_mean)

df_median = df.fillna(df.median())
print("\nDataFrame after replacing missing values with Median:")
print(df_median)

df_dropped = df.dropna()
print("\nDataFrame after dropping rows with missing values:")
print(df_dropped)
```

# Q2) Write a python program using pandas to :
a) Create a dataset containing duplicate rows  b) Identify duplicate entries  c) Remove all duplicate rows from the dataset

```python
import pandas as pd

data = {
    'ID': [1, 2, 3, 2, 4, 3],
    'Name': ['Alice', 'Bob', 'Charlie', 'Bob', 'David', 'Charlie'],
    'Marks': [85, 90, 78, 90, 88, 78]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nDuplicate Rows:")
print(df.duplicated())

print("\nRows that are duplicates:")
print(df[df.duplicated()])

df_no_duplicates = df.drop_duplicates()

print("\nDataFrame after removing duplicate rows:")
print(df_no_duplicates)
```

# Q3) Write a Python program to normalize the following list of values using Min-Max normalization: [12, 25, 40, 55, 70] 
• Use the formula: 𝑥′ = (X−X𝑚𝑖𝑛)/(X𝑚𝑎𝑥−X𝑚𝑖𝑛)

```python
data = [12, 25, 40, 55, 70]

xmin = min(data)
xmax = max(data)

normalized_data = []

for x in data:
    x_norm = (x - xmin) / (xmax - xmin)
    normalized_data.append(x_norm)

print("Original Data:", data)
print("Normalized Data:", normalized_data)
```

# 4) Write a Python program to standardize a dataset using Z-score normalization. The program should: 
• Calculate mean and standard deviation  • Compute Z-score for each value  • Display transformed data 

```python
import numpy as np

data = [12, 25, 40, 55, 70]

mean = np.mean(data)
std_dev = np.std(data)

z_scores = [(x - mean) / std_dev for x in data]

print("Original Data:", data)
print("Mean =", mean)
print("Standard Deviation =", std_dev)

print("\nZ-Score Normalized Data:")
for x, z in zip(data, z_scores):
    print(f"{x} --> {z:.4f}")
```

# 5) Write a Python program to detect outliers in a dataset using the Z-score method. 

```python
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 30, 100]

mean = np.mean(data)
std_dev = np.std(data)

outliers = []

for x in data:
    z_score = (x - mean) / std_dev
    if abs(z_score) > 2:  # Threshold can be 2 or 3
        outliers.append(x)

print("Dataset:", data)
print("Mean =", round(mean, 2))
print("Standard Deviation =", round(std_dev, 2))
print("Outliers:", outliers)
```

# 6) Write a Python program to: 
• Create a dataset containing categorical values such as ["Red", "Blue", "Green", "Blue"]  
• Convert the categorical data into numerical labels using Label Encoding 

```python
from sklearn.preprocessing import LabelEncoder

colors = ["Red", "Blue", "Green", "Blue"]

print("Original Data:")
print(colors)

label_encoder = LabelEncoder()
encoded_colors = label_encoder.fit_transform(colors)

print("\nEncoded Data:")
print(encoded_colors)

print("\nLabel Mapping:")
for color, label in zip(label_encoder.classes_,
                        range(len(label_encoder.classes_))):
    print(f"{color} --> {label}")
```

# 7) Write a Python program to perform One-Hot Encoding on a categorical feature named City containing values: Kolkata, Delhi, Mumbai, Chennai 

```python
import pandas as pd

data = {
    'City': ['Kolkata', 'Delhi', 'Mumbai', 'Chennai']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

encoded_df = pd.get_dummies(df, columns=['City'])

print("\nOne-Hot Encoded Dataset:")
print(encoded_df)
```

# 8) Write a Python program using StandardScaler from Scikit-learn to scale numerical features of a dataset. 

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    'Age': [18, 22, 25, 30, 35],
    'Salary': [20000, 25000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print("\nScaled Dataset:")
print(scaled_df)
```

# 9) Write a Python program to compare: 
• Min-Max Scaling  • Standardization  

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    'Values': [12, 25, 40, 55, 70]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

minmax_scaler = MinMaxScaler()
df['MinMax_Scaled'] = minmax_scaler.fit_transform(df[['Values']])

standard_scaler = StandardScaler()
df['Standardized'] = standard_scaler.fit_transform(df[['Values']])

print("\nDataset after Scaling:")
print(df)
```

# 10) Write a Python program to convert a column containing numeric strings into integer type. 

```python
import pandas as pd

data = {
    'Numbers': ['10', '20', '30', '40', '50']
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nData Type Before Conversion:")
print(df.dtypes)

df['Numbers'] = df['Numbers'].astype(int)

print("\nDataFrame After Conversion:")
print(df)

print("\nData Type After Conversion:")
print(df.dtypes)
```
